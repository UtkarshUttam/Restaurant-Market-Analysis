## Tests and hypotheses
Spearman tests rank association (H0: no monotonic association; H1: a monotonic association exists). Welch t-tests compare mean ratings between service groups (H0: equal means; H1: unequal means). Kruskal-Wallis compares rating distributions across price bands (H0: same distribution; H1: at least one differs). A small p-value does not establish causation or business importance. Cost tests are stratified by currency.
