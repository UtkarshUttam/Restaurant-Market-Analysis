## Score and recommendation boundary
Score = (cuisine mean rating - mean rating of its currency group) x (1 - cuisine restaurant-count percentile within that currency). Only segments with at least five rated restaurants are ranked. This transparent heuristic screens for above-market ratings and lower relative presence; it does not measure demand or predict success. Validate candidate areas with market size, transactions, rent, and customer research before recommending investment.
